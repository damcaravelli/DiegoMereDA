### Exercici 1. Calculadora de l'índex de massa corporal

- Escriu una funció que calculi l'IMC ingressat per l'usuari/ària, és a dir, qui ho executi haurà d'ingressar aquestes dades

- La funció ha de classificar el resultat en les seves respectives categories

```python
def IndiceMasaCorporal(): #Crear la funciona
    Peso = float(input("Ingrese su peso en kg:")) #En este caso pido los datos dentro de la funcion, pero bien se podrian dar como parametros
    Altura = float(input("Ingrese su altura en metros:"))
    IMC = Peso / (Altura**2) #Calculo del IMC, Peso entre altura al cuadrado
    #Bucles para asignar el IMC resultante a una categoria
    if IMC < 18.5:
        categoria = "Bajo Peso"
    elif IMC >= 18.5 and IMC <= 24.9:
        categoria = "Peso normal"
    elif IMC >= 25 and IMC <= 29.9:
        categoria = "Sobrepeso"
    elif IMC > 30:
        categoria = "Obesidad"   

    return [Peso, Altura, IMC, categoria] #La funcion regresa el peso y altura ingresado, asi como los dos calculos, IMC y categoria


Usuario = IndiceMasaCorporal() #Pruebo la funcion y asigno el resultado a una vaiable

#Imprimo el resultado como prueba
print (f"Para su peso de: {Usuario[0]} Y su altura de: {Usuario[1]}, Su indice de masa corporal es: {round(Usuario[2],2)} - {Usuario[3]}")
```

### Exercici 3 Comptador de paraules d'un text.

Escriu una funció que donat un text, mostri les vegades que apareix cada paraula.

```python
def Cuenta_Palabras (Texto):
    Dic = {} #Creo Diccionario vacio

    for caracter in "().,;:!?'\"":
        Texto = Texto.replace(caracter, "") #Reemplazo alguno de los caracteres con vacio para no contarlos en las palabras

    Texto = Texto.lower().split() # Cambio todo el texto a minusculas (para contar las mismas palabras) y separo las palabras en lista

    for palabra in Texto: #For para contar
        if palabra in Dic: #Si la palabra ya esta, sumo 1
            Dic[palabra] += 1
        else:
            Dic[palabra] = 1 #Si la palabra no esta, la creo con 1
    return Dic #Regerso el diccionario

Texto_Contado = Cuenta_Palabras(input("Ingrese un texto para contar sus palabras")) #Pido el texto y lo asigno a una variable

for x, y in Texto_Contado.items():
    print(f"La palabra '{x}' aparece: {y} veces") #Imprimo el diccionario
```

### Exercici 4 Diccionari invers.

Resulta que el client té una enquesta molt antiga que s'emmagatzema en un diccionari i els resultats els necessita al revés, és a dir, intercanviats les claus i els valors. Els valors i claus en el diccionari original són únics; si aquest no és el cas, la funció hauria d'imprimir un missatge d'advertiment.

```python
Dic_Original = {'a':1,'b':2,'c':3} #Diccionario a invertir 

def Invertir_Diccionario(Dic_Original): #Funcion con un diccionario como parametro 
    Nuevo_Dic = {} #Creo el nuevo diccionario
    Set_Valores = set() #Creo un set con los valores que voy invirtiendo para comprobar que no haya repetidos

    for x, y in Dic_Original.items():
        if y in Set_Valores: #Reviso si el valor ya esta, en caso, regreso el mensaje de error 
            return "El diccionario tiene valores repetidos"
        
        else:
            Set_Valores.add(y) #En caso de que no este, lo anado al set de valores para comrpbar 
            Nuevo_Dic[y] = x #Volteo el diccionario
    return Nuevo_Dic


Invertir_Diccionario(Dic_Original)

```

Tingues en compte que si és un valor únic no ha de ser una llista.

```python
Dic_Original2 = {'a':1,'b':2,'c':3, 'd':3, 'e':3} #Diccionario con duplicados

def Invertir_Diccionario_Duplicados(Diccionario):

    Dic_Invertido = {} #Creo diccionario nuevo

    for x,y in Diccionario.items(): #Bucle con valor y llave del diccionario
        if y in Dic_Invertido: #EN caso de que ya este el valor
            if isinstance(Dic_Invertido[y], list):# Compruebo si ya es una lista, si ya lo es, le anado el ultimo valor
                Dic_Invertido[y].append(x)
            else: 
                Dic_Invertido[y] = [Dic_Invertido[y],x] #Si no, creo la lista
        else:
            Dic_Invertido[y] = x #En caso de que no este, lo creo
    return Dic_Invertido

   
Diccionario  = Invertir_Diccionario_Duplicados(Dic_Original2)
Diccionario
```

### Exercici 2 Conversió de tipus de dades

El client rep una llista de dades i necessita generar dues llistes, la primera on estaran tots els elements que es van poder convertir en flotants i l'altra on estan els elements que no es van poder convertir. Exemple de la llista que rep el client: [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]

```python
Lista = [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]

def Convertidor_Datos(Listas): #Funcion que recibe una lista como parametro

    Lista_Flotante = [] #Inicio lista para los numeros flotantes y otra para los que no
    Lista_No_Flotante = []

    for x in Listas: #Recorro la lista
        try:
            if isinstance(x,str): #Si es un string, comrpuebo que se pueda convertir a flotante y lo hago
                x = float(x)

            if isinstance(x,(float,int)): #compruebo que el valor sea entero o flotante (incluyendo el primer valor si se pudo convertir)
                Lista_Flotante.append(float(x)) #Si se puede, lo anado a la lista de flotantes como float
            elif isinstance(x,(list,tuple)): #Compruebo si es una lista o una tupla
                Lista_Flotante_Secundaria, Lista_No_Flotante_Secundaria = Convertidor_Datos(x) #Si lo es, llamo la funcion de forma recursiva 
                Lista_Flotante.extend(Lista_Flotante_Secundaria) #Anado la funcion flotante que retorna (recursiva) a mi flotante 
                Lista_No_Flotante.extend(Lista_No_Flotante_Secundaria) #Igual con la no flotante
            else:
                Lista_No_Flotante.append(x) #Resto de casos, anado a la no flotante
    
        except ValueError: #En caso de fallar la conversion del string a float, lo anado tambien a la lista de no flotantes
            Lista_No_Flotante.append(x)

    return Lista_Flotante, Lista_No_Flotante 

L1, L2 = Convertidor_Datos(Lista)

print(f"La lista de numeros es: '{L1}' \nY la que no se puede convertir: {L2}")

```
